building a chatbot

```python
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage
from langchain_core.prompts import PromptTemplate, FewShotPromptTemplate
from dotenv import load_dotenv
import os
from openai import OpenAI
```

```python
load_dotenv()
api_key=os.getenv("GROQ_API_KEY")
```

```python
class chatbot:
    def __init__(self,name,instructions,examples,model="openai/gpt-oss-20b",temparature=0.0):
        self.llm=ChatOpenAI(
            model=model,
            temperature=temparature,
            api_key=api_key,
            base_url="https://api.groq.com/openai/v1"
        )
        example_prompt = PromptTemplate(
            input_variables=["input", "output"],
            template="Human: {input}\nAI: {output}"
        )
        prompt_template=FewShotPromptTemplate(
            example_prompt = example_prompt,
            examples = examples,
            prefix= instructions,
            suffix= "",
            input_variables= ['']
        )
        self.message= prompt_template.invoke({}).to_messages()
    def invoke(self, user_msg):
        self.message.append(HumanMessage(user_msg))
        ai_msg=self.llm.invoke(self.message)
        self.message.append(ai_msg)
        return ai_msg
```

```python
instructions= (
    "you are a helpful assistant"
    "just give one word short answers"
)
```

```python
examples=[
    {
        "input": "largest country",
        "output": "russia"
    },
    {
        "input": "dense populated country",
        "output": "india"
    }
]
```

```python
beep_chat = chatbot(
    name="GK master",
    instructions=instructions,
    examples=examples
)
```

```python
beep_chat.invoke('what the coolest place')
```

```python
beep_chat.invoke('where can i plan trip on summers')
```
